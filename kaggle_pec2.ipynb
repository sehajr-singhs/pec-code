# PEC2 — property-conditioned world models for contact-rich control

Runs experiments E0–E4 on a Kaggle GPU (T4/P100):

- **E0** blind property encoder vs mean-predictor (identifiability bar)
- **E1** dream-training with the contrastive property gate vs observation noise vs honest dreams
- **E2** latent-ODE vs capacity-matched fixed-tick world model across clocks/horizons
- **E3** mid-episode actuator fault: blind vs posterior-conditioned vs oracle
- **E4** narrow-distribution training, edge-of-distribution evaluation strata

**Setup:** nothing to attach - the pec2 source is embedded below as a base64 zip (self-contained, no internet needed).

```python
import base64, os, sys, zipfile

WORK = "/kaggle/working"
CODE = os.path.join(WORK, "code")
os.makedirs(CODE, exist_ok=True)
B64 = (
    ''
)
with zipfile.ZipFile(__import__('io').BytesIO(base64.b64decode(B64))) as zf:
    zf.extractall(CODE)
sys.path.insert(0, CODE)
print("code root:", CODE)
print("contents:", sorted(os.listdir(CODE))[:12])
```

```python
import torch
print("torch", torch.__version__, "| cuda:", torch.cuda.is_available(),
      "|", (torch.cuda.get_device_name(0) if torch.cuda.is_available() else "cpu"))
from pec2.envs import PushWorld
w = PushWorld(n=4096, device="cuda" if torch.cuda.is_available() else "cpu")
w.reset()
o, r, d = w.act(torch.zeros(w.n, 3, device=w.device))
print("env smoke:", tuple(o.shape), float(r.mean()))
```

```python
import json
s = json.load(open(f"{WORK}/results_full/stats_pec2.json"))
print("E0 encoder NRMSE", round(s["e0"]["mean_nrmse"], 3),
      "vs mean-predictor", round(s["e0"]["mean_predictor_nrmse"], 3))
print("E1 gate vs noise:", s["e1"]["stats"]["gate_vs_noise"])
print("E2 latent-ODE MSE:", json.dumps(s["e2"]["mse"]["latent_ode"]))
print("E2 fixed-tick MSE:", json.dumps(s["e2"]["mse"]["discrete_tick"]))
print("E3 zcond vs blind:", s["e3"]["stats"]["zcond_vs_blind"])
print("E3 oracle vs zcond:", s["e3"]["stats"]["oracle_vs_zcond"])
print("E4 edge:", s["e4"]["stats"]["zcond_vs_blind_edge"])
print("E4 double-edge:", s["e4"]["stats"]["zcond_vs_blind_double"])
```

```python
import os
from IPython.display import Image, display
figdir = f"{WORK}/results_full/figures"
if os.path.isdir(figdir):
    for f in sorted(os.listdir(figdir)):
        if f.endswith(".png"):
            print("--", f)
            display(Image(os.path.join(figdir, f)))
```

```python
out = "/kaggle/working/pec2_artifacts.zip"
with zipfile.ZipFile(out, "w", zipfile.ZIP_DEFLATED) as zf:
    for root in (f"{WORK}/results_full", f"{WORK}/results_smoke"):
        if os.path.isdir(root):
            for dirpath, _, files in os.walk(root):
                for f in files:
                    p = os.path.join(dirpath, f)
                    zf.write(p, os.path.relpath(p, WORK))
print("wrote", out)
```
